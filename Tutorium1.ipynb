#### Aufgabe 1 a) 
Führen Sie Zeile für Zeile den Code aus und überlegen Sie sich jeweils vorher was die Zeile macht. Lassen 
Sie sich den Text am Ende der Aufgabe ausgeben und vollziehen Sie nach, wie der Text entsteht.

```python
import random
```

```python
adjective_options = ['riesengroß', 'flauschig', 'steinhart', 'glibberig', 'matschig']
animal_options = ['Blattlaus', 'Waldkatze', 'Wüstenrennmaus']
place_options = ['TU Campus', 'Charlottenburg', 'Alexanderplatz']
food_options = ['Hummus-Dip', 'Steak']
substance_options = ['Blausäure', 'Algenessig', 'Granit']
job_options = ['Chemielaborant', 'Verwaltungsfachangestellter']
```

```python
random.shuffle(adjective_options)

adjective1 = adjective_options[0]
adjective2 = adjective_options[1]
adjective3 = adjective_options[2]
adjective4 = adjective_options[3]
```

```python
random.shuffle(animal_options)
animals_character1 = animal_options[0]
animals_character2 = animal_options[1]
animal = animal_options[2]
```

```python
random.shuffle(food_options)
food1 = food_options[0]
food2 = food_options[1]
```

```python
place = random.choice(place_options)
substance = random.choice(substance_options)
job = random.choice(job_options)
```

```python
text = 'Eine ' + adjective1 + 'e ' + animals_character1 + ' trifft ihren Kumpel, eine ' + adjective2 + 'e ' + animals_character2 + \
    ', in einem ' + adjective3 + 'en, schon lange verlassenen ' + place + '. Die ' + animals_character1 + \
    ' sagt: "Hier, probier mal meine neueste Getränkekreation, ein Shake aus ' + food1 + ' und ' + food2 + \
    '!" Der Kumpel trinkt, verzieht angewidert das Gesicht und murmelt leise: "' + adjective4 + '!" Laut sagt er: "Bääh, das schmeckt ja wie ' + \
    animal + ' mit ' + substance + '! Also Koch bist du nicht, höchstens ' + job + '!"'
print(text)
```

#### Aufgabe 1 b)
Definieren Sie selbst eigene Variablen, wählen Sie zufällig ein Element aus mehreren Möglichkeiten und fügen Ihre Auswahl in den Text ein.
