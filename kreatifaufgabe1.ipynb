```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse

fig, ax = plt.subplots(figsize=(6.5, 6.5))
ax.set_xlim(0, 10)
ax.set_ylim(0.5, 10.05)
fig.patch.set_facecolor("#2E052E")

circle1 = plt.Circle((5, 5), 3, color="#FFFFE2")
circle2 = plt.Circle((5, 5), 2.5, color="#F3B353")
circle3 = plt.Circle((5, 5), 1.8, color="#FAF768")
oval1 = Ellipse((3.5, 6.2), width=1, height=1.3, angle=-45, color='#FFFFE2')
oval2 = Ellipse((3, 5), width=0.6, height=0.5, angle=60, color='#FFFFE2')
thing = Ellipse((3.7, 2.5), width=2, height=1, angle=-30, color="#94788CFF")
thing2 = Ellipse((3.5, 2.2), width=2, height=1, angle=-30, color="#63465AFF")
floor = Ellipse((5, 1.4), width=7, height=1, color="#C7837AFF")
floor2 = plt.Rectangle((1, 0), 8, 1.2, color="#2E052E")
bg = plt.Circle((5, 5), 5, color="#832F3D94")

ax.add_patch(bg)
ax.add_patch(floor2)
ax.add_patch(floor)
ax.add_patch(thing2)
ax.add_patch(thing)
ax.add_patch(circle1)
ax.add_patch(circle2)
ax.add_patch(circle3)
ax.add_patch(oval1)
ax.add_patch(oval2)


ax.axis('off')
plt.title("Die Glühbirne ist die Sonne", color="#FFFFE2", x=0.5, y=0.85)
plt.show()
```
